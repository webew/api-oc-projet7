# Test en local

## /health

```python
import requests

BASE_URL = "http://127.0.0.1:8000"

r = requests.get(f"{BASE_URL}/health", timeout=10)
print("status_code:", r.status_code)
print("json:", r.json())
```

## /predict

```python
import requests

BASE_URL = "http://127.0.0.1:8000"

# On prend un SK_ID_CURR réel depuis application_test.csv
import pandas as pd
df = pd.read_csv("data/raw/application_test.csv")
sk_id = int(df["SK_ID_CURR"].iloc[0])

payload = {"sk_id_curr": sk_id}

r = requests.post(f"{BASE_URL}/predict", json=payload, timeout=30)
print("status_code:", r.status_code)
print("json:", r.json())
```

## /predict

```python
import requests
import pandas as pd

BASE_URL = "https://api-oc-projet7.onrender.com"

# Test /predict
df = pd.read_csv("data/raw/application_test.csv")
sk_id = int(df["SK_ID_CURR"].iloc[0])

payload = {"sk_id_curr": sk_id}

r = requests.post(f"{BASE_URL}/predict", json=payload, timeout=60)
print("\npredict - status_code:", r.status_code)
print("predict - json:", r.json())
```
